# Just Train Twice (JTT) — CelebA

## 4.1 Overview

**Just Train Twice (JTT)** is a group-robust learning method designed to improve model performance under spurious correlations without requiring group annotations during training.

JTT uses a two-stage training procedure:

1. **Error identification:** Train an initial ERM model for a small number of epochs and identify training examples it misclassifies.
2. **Error upweighting:** Train a new model from scratch, assigning greater weight to the previously misclassified examples.

The intuition is that misclassified examples often include minority-group samples that contradict spurious correlations learned by the model. Upweighting these examples encourages the classifier to learn more robust features.

**Our experiment:** Using CelebA, we predict `Blond_Hair` with `Male` as the spurious attribute. We evaluate whether JTT improves worst-group accuracy compared with standard ERM and our gradient-based filtering methods, using the same pretrained ResNet-18 features.
